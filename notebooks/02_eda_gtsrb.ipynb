# EDA — German Traffic Sign Recognition Benchmark (GTSRB)

Exploración del dataset para entender distribución de clases, tamaño de imágenes y calidad de datos.

```python
import sys
sys.path.insert(0, '..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from PIL import Image
from collections import Counter

from src.config import LABEL_MAP

# Ajusta esta ruta al directorio donde descargaste el dataset
TRAIN_DIR = Path('../data/gtsrb/Train')
```

```python
# Contar imágenes por clase
class_counts = {}
for cls_dir in sorted(TRAIN_DIR.iterdir()):
    if cls_dir.is_dir():
        class_counts[int(cls_dir.name)] = len(list(cls_dir.glob('*.png')))

print(f'Total clases: {len(class_counts)}')
print(f'Imágenes totales: {sum(class_counts.values())}')
print(f'Clase con más imágenes: {max(class_counts, key=class_counts.get)} ({max(class_counts.values())} imágenes)')
print(f'Clase con menos imágenes: {min(class_counts, key=class_counts.get)} ({min(class_counts.values())} imágenes)')
```

```python
# Distribución de clases
sorted_counts = sorted(class_counts.items(), key=lambda x: x[1], reverse=True)
cids = [str(c[0]) for c in sorted_counts]
counts = [c[1] for c in sorted_counts]

plt.figure(figsize=(18, 5))
plt.bar(cids, counts, color='steelblue')
plt.xlabel('Clase (ID)')
plt.ylabel('Número de imágenes')
plt.title('Distribución de imágenes por clase — GTSRB')
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig('../reports/figures/distribucion_clases.png', dpi=150)
plt.show()
```

```python
# Ejemplos visuales: una imagen por clase
fig, axes = plt.subplots(7, 7, figsize=(16, 16))
axes = axes.flatten()

for cls_id in range(43):
    cls_dir = TRAIN_DIR / str(cls_id)
    img_paths = list(cls_dir.glob('*.png'))
    if img_paths:
        img = Image.open(img_paths[0]).convert('RGB').resize((64, 64))
        axes[cls_id].imshow(img)
        axes[cls_id].set_title(f'{cls_id}', fontsize=8)
        axes[cls_id].axis('off')

# Apagar ejes sobrantes
for i in range(43, len(axes)):
    axes[i].axis('off')

plt.suptitle('Ejemplos por clase — GTSRB', fontsize=14)
plt.tight_layout()
plt.savefig('../reports/figures/ejemplos_por_clase.png', dpi=150)
plt.show()
```
